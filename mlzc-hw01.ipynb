```python
import pandas as pd
import numpy as np

# Load dataset
url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"

df = pd.read_csv(url)

# Q1 - Pandas version
print("Q1:", pd.__version__)

# Q2 - Number of records
print("Q2:", len(df))

# Q3 - Number of fuel types
print("Q3:", df["fuel_type"].nunique())

# Q4 - Number of columns with missing values
print("Q4:", df.isnull().any().sum())

# Q5 - Maximum fuel efficiency of cars from Asia
print("Q5:", df[df["origin"] == "Asia"]["fuel_efficiency_mpg"].max())

# Q6 - Median horsepower before and after filling missing values
median_before = df["horsepower"].median()

most_frequent = df["horsepower"].mode()[0]

df["horsepower"] = df["horsepower"].fillna(most_frequent)

median_after = df["horsepower"].median()

print("Q6 - Median before:", median_before)
print("Q6 - Median after:", median_after)

if median_after > median_before:
    print("Q6 answer: Yes, it increased")
elif median_after < median_before:
    print("Q6 answer: Yes, it decreased")
else:
    print("Q6 answer: No")

# Q7
asia = df[df["origin"] == "Asia"]

asia = asia[["vehicle_weight", "model_year"]]

X = asia.iloc[:7, :].to_numpy()

XT = X.T

XTX = XT @ X

XTX_inverse = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inverse @ XT @ y

print("Q7 weights:", w)
print("Q7 sum:", w.sum())
```
